# `crm_sales_details` Exploratory Data Analysis

## Init

```python
from pyspark.sql.functions import col, try_to_date
```

## Reading from Bronze layer

```python
df = spark.table("data_lakehouse.bronze.crm_sales_details")
```

## Exploring the data

### Incorrect date values

```python
df.where(try_to_date(col("sls_order_dt"), "yyyyMMdd").isNull()).show()
```

### Incorrect `sls_sales` values

```python
df.where(
    (col("sls_sales").isNull()) |
    (col("sls_sales").cast("int") <= 0) |
    (col("sls_sales").cast("int") != col("sls_quantity").cast("int") * col("sls_price").cast("int"))
).show()
```

### Incorrect `sls_price` values

```python
df.where(
    (col("sls_price").isNull()) |
    (col("sls_price").cast("int") <= 0)
).show()
```

## Issues found

1. Column: `sls_order_dt` has invalid values.
2. Column: `sls_sales` has incorrect values.
3. Column: `sls_price` has incorrect values.
4. Columns must be casted to their respective data types (e.g. `sls_order_dt` to `date`, `sls_sales` to `decimal`).
5. Column names do not follow any naming convention (e.g. `sls_ord_num` instead of `order_number`, `sls_ship_dt` instead of `ship_date`).
