# `crm_cust_info` Exploratory Data Analysis

## Init

```python
from pyspark.sql.functions import col, length, trim
```

## Reading from Bronze layer

```python
df = spark.table("data_lakehouse.bronze.crm_cust_info")
```

## Exploring the data

### Primary key null values

```python
df.where(col("cst_id").isNull()).show()
```

> There are null rows

### Primary key non-unique values

```python
df.groupBy(col("cst_id")).count().where(col("count") > 1).show()
```

### Leading and trailing spaces

```python
df.where(length(col("cst_firstname")) != length(trim(col("cst_firstname")))).show()
```

```python
df.where(length(col("cst_lastname")) != length(trim(col("cst_lastname")))).show()
```

### Non-descriptive values

```python
df.select(col("cst_marital_status")).distinct().show()
```

```python
df.select(col("cst_gndr")).distinct().show()
```

## Issues found

1. Column: `cst_id` (primary key) has null values.
2. Column: `cst_id` (primary key) has non-unique values.
3. Columns: `cst_firstname` and `cst_lastname` have leading and trailing spaces.
4. Columns: `cst_marital_status` and `cst_gndr` do not have descriptive values.
5. Columns must be casted to their respective data types (e.g. `cst_id` to `bigint`, `cst_create_date` to `date`).
6. Column names do not follow any naming convention (e.g. `cst_id` instead of `customer_id`, `cst_gndr` instead of `gender`).
