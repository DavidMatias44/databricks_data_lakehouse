# `erp_loc_a101` Exploratory Data Analysis

## Init

```python
from pyspark.sql.functions import col
```

## Reading from Bronze layer

```python
df = spark.table("lakehouse_project.bronze.erp_loc_a101")
```

## Exploring the data

### Unnecesary `-` in `CID` column

```python
df.select(col("CID")).show(10)
```

### Non-consistent values

```python
df.select(col("CNTRY")).distinct().show()
```

## Issues found

1. Column: `CID` has prefix character '-' which is not necessary.
2. Column: `CNTRY` has non-consistent values.
3. Columns must be casted to their respective data types (e.g. `CID` to `bigint`).
4. Column names do not follow any naming convention (e.g. `CID` instead of `customer_id`, `CNTRY` instead of `country`).
