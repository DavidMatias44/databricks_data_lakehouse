# `erp_cust_az12` Exploratory Data Analysis

## Init

```python
from pyspark.sql.functions import col, current_date
```

## Reading from Bronze layer

```python
df = spark.table("lakehouse_project.bronze.erp_cust_az12")
```

## Exploring the data

### Prefix 'nas' in column `cid` is not necessary

```python
df.select(col("CID")).show(10)
```

### Invalid date values

```python
df.where(
    (col("BDATE") >= current_date()) |
    (col("BDATE") <= "1925-01-01")
).show()
```

### Non-descriptive and non-consistent values

```python
df.select(col("GEN")).distinct().show()
```

## Issues found

1. Column: `CID` has prefix 'NAS' which is not necessary.
2. Column: `BDATE` has invalid date values.
3. Column: `GEN` has non-descriptive and non-consistent values.
4. Columns must be casted to their respective data types (e.g. `CID` to `bigint`, `BDATE` to `date`).
5. Column names do not follow any naming convention (e.g. `CID` instead of `customer_id`, `BDATE` instead of `birth_date`).
